## Problem 1: BFS vs. DFS Implementation and Analysis

Implement both **Breacth-First Search (BFS)** and **Depth-First Search (DFS)** in **Python** for the examples introduced in **Lecture - Search(1)**.

* Find a path from A to E using BFS in graph.
* Find a path from A to E using DFS in graph.
* Find a path from A to B using BFS in grid.
* Find a path from A to B using DFS in grid.

Compare the results and write your findings on the differences between BFS and DFS (e.g., traversal order, memory usage, path length).



```python
# Graph

class Node:
    def __init__(self, name):
        self.name = name
        self.children = []

def makeGraph():
    A = Node('A')
    B = Node('B')
    C = Node('C')
    D = Node('D')
    E = Node('E')
    F = Node('F')
    A.children.append(B)
    B.children.append(C)
    B.children.append(D)
    D.children.append(F)
    C.children.append(E)
    return A

graph = makeGraph()
```

```python
# Grid 

grid_text = """_#####
______
_##_##
_##_##
_#__##
_#_###
___###
"""

class Grid:
    def __init__(self, width, height, text):
        self.width = width
        self.height = height
        self.are_floors = [False for _ in range(width * height)]
        cell_i = 0
        for c in text:
            if c == '#':
                cell_i += 1
            if c == '_':
                self.are_floors[cell_i] = True
                cell_i += 1
                
    def getIsFloor(self, point):
        if point[0] < 0:
            return False
        if point[0] > self.width - 1:
            return False
        if point[1] < 0:
            return False
        if point[1] > self.height - 1:
            return False
        return self.are_floors[point[0] + point[1] * self.width]
                

grid = Grid(6, 7, grid_text)
```

```python
# Problem 1 Part a

from collections import deque

def graphBfs(start_node, goal_name):
    print(f"performing graph bfs from {start_node.name} to {goal_name}")
    frontier = deque([(start_node, [start_node])]) # queue
    explored = set()
    while frontier:
        node, path = frontier.popleft()
        if node in explored:
            continue
        explored.add(node)
        print(f"explored {node.name}")
        if node.name == goal_name:
            print(f"final path: {[vertex.name for vertex in path]}")
            return
        for child in node.children:
            if child not in explored:
                frontier.append((child, path + [child]))
    print("found no path to goal")

graphBfs(graph, 'E')
```

```python
# Problem 1 Part B

def graphDfs(start_node, goal_name):
    print(f"performing graph dfs from {start_node.name} to {goal_name}")
    frontier = [(start_node, [start_node])] # stack
    explored = set()
    while frontier:
        node, path = frontier.pop()
        if node in explored:
            continue
        explored.add(node)
        print(f"expored {node.name}")
        if node.name == goal_name:
            print(f"final path: {[vertex.name for vertex in path]}")
            return
        for child in node.children:
            if child not in explored:
                frontier.append((child, path + [child]))
    print("found not path to goal")

graphDfs(graph, 'E')
```

```python
# Problem 1 Part C

from collections import deque

def gridBfs(grid, start, goal):
    print(f"performing grid bfs from {start} to {goal}")
    frontier = deque([(start, [start])]) # queue
    explored = set()
    while frontier:
        point, path = frontier.popleft()
        if point in explored:
            continue
        explored.add(point)
        print(f"explored {point}")
        if point[0] == goal[0] and point[1] == goal[1]:
            print(f"final path: {[pair for pair in path]}")
            return
        left = (point[0]-1, point[1])
        if left not in explored and grid.getIsFloor(left):
            frontier.append((left, path + [left]))
        right = (point[0]+1, point[1])
        if right not in explored and grid.getIsFloor(right):
            frontier.append((right, path + [right]))
        up = (point[0], point[1]-1)
        if up not in explored and grid.getIsFloor(up):
            frontier.append((up, path + [up]))
        down = (point[0], point[1]+1)
        if down not in explored and grid.getIsFloor(down):
            frontier.append((down, path + [down]))
    print("path not found to goal")

gridBfs(grid, (0, 6), (3, 3))
```

```python
# Problem 1 part D

def gridDfs(grid, start, goal):
    print(f"performing grid dfs from {start} to {goal}")
    frontier = [(start, [start])] # stack
    explored = set()
    while frontier:
        point, path = frontier.pop()
        if point in explored:
            continue
        explored.add(point)
        print(f"explored {point}")
        if point[0] == goal[0] and point[1] == goal[1]:
            print(f"final path: {[pair for pair in path]}")
            return
        left = (point[0]-1, point[1])
        if left not in explored and grid.getIsFloor(left):
            frontier.append((left, path + [left]))
        right = (point[0]+1, point[1])
        if right not in explored and grid.getIsFloor(right):
            frontier.append((right, path + [right]))
        up = (point[0], point[1]-1)
        if up not in explored and grid.getIsFloor(up):
            frontier.append((up, path + [up]))
        down = (point[0], point[1]+1)
        if down not in explored and grid.getIsFloor(down):
            frontier.append((down, path + [down]))
    print("path not found to goal")

gridDfs(grid, (0, 6), (3, 3))
```

### Problem 1 Conclusion

It is clear that both DFS and BFS have different use cases. BFS involves searching the vertices of a graph ordered by distance from the start vertex, while DFS searches the nodes of a graph immediatly. BFS has higher memory usage, because all unexplored adjacent vertices to each vertex removed form the queue needs to be added to the queue, while DFS can be more memory efficient because it is possible to implement the algorithm using recursive functions instead of a data structure to store the frontier. 

It appears that the path length result of BFS can sometimes be longer than that of DFS. BFS always gives the shortest path between the start and goal vertices, while DFS may choose a longer path. However, DFS appears to sometimes be faster at determining a viable path in cases where the start and goal nodes are far apart and there are vertices that do not exit in a path from the start to the goal.

## Problem 2: DLS on the grid.

Using the provided above grid (start **A**, goal **B**), **implement Depth-Limited Search in Python** and run it with two depth limits.

Run DLS with **limit = 6** and **limit = 10** on the grid.

Compare the two runs. Discuss how changing the depth limit impacts success/completeness, node expansions, and path quality. Briefly contrast DLS with your BFS/DFS results.

```python
# Problem 2

def gridDls(grid, start, goal, limit):
    print(f"performing grid dls from {start} to {goal} with limit {limit}")
    frontier = [(start, [start], 0)] # stack
    explored = set()
    while frontier:
        point, path, depth = frontier.pop()
        if point in explored:
            continue
        if depth >= limit:
            continue
        explored.add(point)
        print(f"explored {point}")
        if point[0] == goal[0] and point[1] == goal[1]:
            print(f"final path: {[pair for pair in path]}")
            return
        left = (point[0]-1, point[1])
        if left not in explored and grid.getIsFloor(left):
            frontier.append((left, path + [left], depth + 1))
        right = (point[0]+1, point[1])
        if right not in explored and grid.getIsFloor(right):
            frontier.append((right, path + [right], depth + 1))
        up = (point[0], point[1]-1)
        if up not in explored and grid.getIsFloor(up):
            frontier.append((up, path + [up], depth + 1))
        down = (point[0], point[1]+1)
        if down not in explored and grid.getIsFloor(down):
            frontier.append((down, path + [down], depth + 1))
    print("path not found to goal")

gridDls(grid, (0, 6), (3, 3), 6)
```

```python
gridDls(grid, (0, 6), (3, 3), 10)
```

### Problem 2 Conclusion

The first run failed to find a solution because there are no paths that connect the start and goal nodes that have a length less than or equal to 6. This is because a DLS can only find a solution if the limit is greater than or equal to the length of the shortest viable path. Meanwhile, the second run was successful because the distance limit of 10 is within the range between the two nodes. However, the second run first attempoted to discover the goal node from the start node through a path that was shorter than 10, so it wasted time exploring a path that could not reach it. This shows that if the limit of a DLS is too large, the algorithm will waste time.
